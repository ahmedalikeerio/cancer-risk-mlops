```python
plt.figure(figsize=(5,5))

sns.countplot(data=df, x='Risk_Level')
plt.title("Risk level Distribution")
plt.xlabel("Risk_Level")
plt.ylabel("Number of Patients")
plt.tight_layout()
plt.show()
```

```python
categorical_features= df.select_dtypes(include='object').columns.to_list()
categorical_features
```

```python
for feature in categorical_features:
    if feature == 'Patient_ID':
        continue
    else:
        print(f"\n {'=' * 50}")
        print(feature)
        print(df[feature].value_counts())
```

```python
numerical_features=df.select_dtypes(include=['int64','float64']).columns.to_list()
numerical_features
```

```python
df[numerical_features].describe().T
```

Feature Relation and Correlation

```python
correlation = df[numerical_features].corr()

plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x='Risk_Level',y='Age')
plt.title("Risk Level vs Age")
plt.xlabel("Risk Level")
plt.ylabel("Age")
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Risk_Level",
    y="BMI"
)

plt.title("BMI Distribution by Risk Level")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Risk_Level",
    y="Smoking"
)

plt.title("Smoking Score by Risk Level")
plt.tight_layout()
plt.show()
```

```python
# Family History

pd.crosstab(
    df["Family_History"],
    df["Risk_Level"],
    normalize="index"
) * 100
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Family_History",
    hue="Risk_Level"
)

plt.title("Family History vs Risk Level")
plt.tight_layout()
plt.show()
```
